## Missing values, Negative values and duplicates in the dataset.

```python
txn_df.shape
finance = txn_df.copy()
```

## null value check in the dataset: 

```python
finance.isna().sum()
```

```python
condition1 = (finance['actual_amount'].isna())
missing_actual = finance[condition1]
print(missing_actual.shape)
display(missing_actual)
```

Negative values check in the dataset:

```python
negative_actual = finance[finance['actual_amount'] < 0]
negative_budget = finance[finance['budget_amount'] < 0]

print(negative_actual.shape)
display(negative_actual)

print("Negative budget amount records: ", negative_budget.shape)
display(negative_budget)
```

Duplicate values check in the dataset:

```python
finance.duplicated().sum()
```

```python
finance_clean = finance.drop_duplicates().copy()
```

```python
finance_clean.to_csv('finance_clean.csv', index=False)
```

### Methods we used above: 
- `.isna()` marks missing values as True and non-missing values as False.
- `.isna().sum()` counts the number of missing values in each column.
- `.isnull()` marks missing values as True and non-missing values as False.
- `< 0` checks for negative values in the dataframe.
- `.duplicated()` marks duplicate rows as True and non-duplicate rows as False.
- `.drop_duplicates()` removes duplicate rows from the dataframe.
- `.copy()` creates a separate copy of the dataframe to avoid modifying the original dataframe.
