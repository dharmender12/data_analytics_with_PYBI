### Variables: 
"Suppose salary is 65,000 and rent is 18000, and EMI is 12000. Let's store the amounts and calculate what remains after the listed expenses."

```python
salary = 65000
rent = 18000
emi = 12000
amount_left = salary - rent - emi 

print("Amount Left:", amount_left)
```
