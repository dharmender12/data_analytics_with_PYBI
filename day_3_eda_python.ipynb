## Dictionary: 
In python, a dictionary is a collection of key-value pairs. Each key is unique and maps to a value. Dictionaries are mutable, meaning you can change their content without changing their identity.

````python
transactions = [{'id':'T101','department':'Travel','amount': 1200,'status':'Approved'},
                {'id':'T102','department':'Travel','amount': 5700,'status':'Approved'},
                {'id':'T103','department':'Travel','amount': 14500,'status':'Approved'},
                {'id':'T104','department':'Travel','amount': 950,'status':'Approved'},
                {'id':'T105','department':'Travel','amount': 22000,'status':'Pending'}
            ]

total = 0
id_x = input("Enter transaction id: ")
for txn in transactions: 
    if txn['id'] > id_x:
        # print(txn['amount'])
        total += txn['amount']

print("Total for transaction id ",id_x," is ",total)
````
