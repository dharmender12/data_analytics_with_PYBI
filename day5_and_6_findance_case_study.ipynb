Duplicate values check in the dataset:

````python
finance.duplicated().sum()
````

````python
finance_clean = finance.drop_duplicates().copy()
````

````python
valid = finance_clean[finance_clean['actual_amount'].notna() 
                       & (finance_clean['actual_amount'].ge(0))].copy() # .ge(0) means greater than or equal to 0 or >= is same
````

````python
missing_actual = finance_clean[finance_clean['actual_amount'].isna()]
negative_actual = finance_clean[finance_clean['actual_amount'] < 0]
````

````python
exceptions = pd.concat([missing_actual, negative_actual])
print("Deduplicated values: ", len(finance_clean))
print("Valid: ", len(valid))
print("Exceptions: ", len(exceptions))
````

````python
valid['variance'] = valid['actual_amount'] - valid['budget_amount']
valid['variance_pct'] = (valid['variance'] / valid['budget_amount']) * 100
````

````python
valid[["transaction_id",
       "budget_amount",
       "actual_amount",
       "variance",
       "variance_pct"]].sort_values(by = ["actual_amount","budget_amount"],ascending=False).head()
````

````python
department_summary = valid.groupby("department").agg(transaction_count = ("transaction_id","count"),
                                                     budget_total = ('budget_amount','sum'),
                                                     actual_total = ('actual_amount','sum'),
                                                     variance_total = ('variance','sum'),
                                                     average_actual = ('actual_amount','mean'),
                                                     average_budget = ('budget_amount','mean')).sort_values(by = 'actual_total',ascending=False)
display(department_summary)
````

### Does the increase in the allocated budget corresponds to an increase in actual expenditure ? 
Pearson correlation is by default correlation in python when you are writing `.corr()`. 
- Its value should be between -1 to 1. 
- If the value is 0 then it shows that there is no relation between the data/column data. 
- If the value is 1 then it shows that there is a perfect  positive correlation between the data or columns.
- If the value is -1 then it shows that there is a perfect negative correlation between the data or columns.
The correlation formula is : 
``` 
    correlation = covariance(x,y) / (std(x) * std(y))

    covariance(x,y) = sum((x - mean(x)) * (y - mean(y))) / n
```


````python
correlation = valid['budget_amount'].corr(valid['actual_amount'])
print("correlation: ", correlation)
````

## Finance risk identification  : 
Overspending Flag : 

````python
valid["spending_status"] = valid.apply(
    lambda row: "Over Budget"
    if row["actual_amount"] > row["budget_amount"]
    else "Within Budget", 
    axis = 1
)

valid[["department","budget_amount","actual_amount","spending_status"]].head()
````

### Calculate the percentage of transactions that exceeds the budgets.

````python
over_budget_count = (valid['spending_status'] == 'Over Budget').sum()
total_txn = len(valid)

over_budget_pct = (over_budget_count / total_txn) * 100
print("Over budget Transactions : ", over_budget_count)
print("Over-budget percentage: ", round(over_budget_pct,2))
````

````python
import numpy as np 

x = valid['budget_amount']
y = valid['actual_amount']

## calculate correlation 
corr = x.corr(y)

## Regression line 
slope, intercept = np.polyfit(x,y,1)
regression_line = slope * x + intercept


## Plot 
plt.figure(figsize = (10,6))

plt.scatter(x, y, alpha=0.7,
            label="Transactions")

plt.plot(x,regression_line,
         color='red',
         label = "Regression Trendline")

plt.xlabel("Budget Amount")
plt.ylabel("Actual Amount")
plt.title(f"Budget vs Actual Expenditure : (r = {correlation:.2f})")

plt.legend()
plt.tight_layout()
plt.show()

````
